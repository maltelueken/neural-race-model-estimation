```python
import numpy as np

from rdmc import rdmc_experiment_simple
from plots import create_delta_plot_rdmc, create_pushforward_plot_rdmc
```

```python
def rdm_experiment_simple_conditions(**kwargs):
    data_congruent = rdmc_experiment_simple(
        **kwargs
    )

    kwargs["amp"] = -kwargs["amp"]

    data_incongruent = rdmc_experiment_simple(
        **kwargs
    )

    condition = np.repeat([1, 0], data_congruent["x"].shape[0])

    return np.c_[np.vstack([data_congruent["x"], data_incongruent["x"]]), condition]
```

## Delta functions

```python
num_trials = 100_000

params_congruent = dict(
    v_c_intercept=0.1,
    v_c_slope=0.5,
    amp=20,
    tau=30,
    s_true=4,
    s_false=4,
    b=80,
    t0=300,
    a_shape=2,
    num_obs=num_trials//2,
    t_max=1000,
    seed=2025
)

data = rdm_experiment_simple_conditions(**params_congruent)

_, ax = create_delta_plot_rdmc(data)

# ax.set_ylim((-30, 50))
```
